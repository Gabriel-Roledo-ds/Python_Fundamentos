# Tabela Verdade com Python

Este notebook explora os operadores lógicos da lógica booleana (E, OU, NÃO, entre outros)
e mostra como eles se comportam através de exemplos práticos em Python.

## O que é uma Tabela Verdade?

Uma tabela verdade é uma representação que mostra todos os possíveis resultados de uma
expressão lógica, combinando os valores **True** (Verdadeiro) e **False** (Falso).

Ela é fundamental para entender:
- Como estruturas condicionais (`if`, `elif`, `else`) avaliam expressões
- Como operadores lógicos (`and`, `or`, `not`) e operadores bit a bit (`&`, `|`, `^`) se comportam
- A base da lógica booleana, usada em programação, eletrônica digital e ciência da computação

## Operadores que vamos explorar

| Operador | Nome    | Descrição                                          |
|----------|---------|-----------------------------------------------------|
| `and` / `&` | E (Conjunção) | Verdadeiro somente se ambos os valores forem True |
| `or` / `\|`  | OU (Disjunção) | Verdadeiro se pelo menos um dos valores for True  |
| `not`       | NÃO (Negação)  | Inverte o valor lógico                            |
| `^`         | XOR (OU exclusivo) | Verdadeiro se os valores forem diferentes     |



## Operador E (AND)

O operador **E** retorna `True` somente quando **ambos** os valores são `True`.
Em Python, pode ser usado com a palavra-chave `and` ou com o operador bit a bit `&`.

````python
print('Tabela verdade E Lógico')
print('************************')
print('True e True =', True & True)
print('True e False =', True & False)
print('False e True =', False & True)
print('False e False =', False & False)
````

##Operador OU (OR)

O operador **OU** retorna `True` quando **pelo menos um** dos valores é `True`.
Só resulta em `False` quando ambos os valores são `False`.
Em Python, pode ser usado com a palavra-chave `or` ou com o operador bit a bit `|`.

````python
print('Tabela verdade OU Lógico')
print('************************')
print('True ou True =', True | True)
print('True ou False =', True | False)
print('False ou True =', False | True)
print('False ou False =', False | False)
````

## Operador NÃO (NOT)

O operador **NÃO** inverte o valor lógico: transforma `True` em `False` e vice-versa.
É o único operador lógico que atua sobre um único valor (operador unário).

````python
print('Tabela verdade NÃO Lógico')
print('************************')
print('não True =', not True)
print('não False =', not False)
````

## Operador XOR (OU Exclusivo)

O operador **XOR** (OU exclusivo) retorna `True` somente quando os dois valores são
**diferentes** entre si. Se forem iguais (ambos True ou ambos False), o resultado é `False`.
Em Python, é representado pelo operador `^`.

````python
print('Tabela verdade XOR (OU Exclusivo)')
print('************************')
print('True xor True =', True ^ True)
print('True xor False =', True ^ False)
print('False xor True =', False ^ True)
print('False xor False =', False ^ False)
````

## Diferença entre `and`/`or` e `&`/`|`



Vale notar que `and`/`or` e `&`/`|` costumam dar o mesmo resultado com booleanos,
mas não são idênticos:

- `and` / `or` fazem **avaliação de curto-circuito** (short-circuit) e avaliam a
  "veracidade" dos valores — funcionam com qualquer tipo de objeto, não só booleanos.
- `&` / `|` são operadores **bit a bit** (bitwise): com booleanos dão o mesmo resultado,
  mas com números inteiros eles comparam a representação **binária** dos números,
  bit por bit.

### Como funciona a operação bit a bit

Veja o caso de `5 & 3` e `5 | 3`:

- `5` em binário é `101`
- `3` em binário é `011`

**`&` (AND bit a bit)** mantém o bit `1` somente onde **ambos** têm `1`:

101 (5) n\
& 011 (3)

001 → 1


**`|` (OR bit a bit)** mantém o bit `1` onde **pelo menos um** dos dois tem `1`:

101 (5) n\
| 011 (3)

111 → 7


Ou seja: `&` e `|` não são exclusivos de booleanos — funcionam com qualquer inteiro,
fazendo essa comparação bit a bit. Já `and`/`or` são pensados para lógica booleana
(embora também funcionem com outros tipos, de um jeito diferente).

Veja um exemplo prático dessa diferença abaixo.

````python
print('Comparando and/or com &/|')
print('************************')
print('True and False =', True and False)
print('True & False   =', True & False)
print()
print('5 & 3 =', 5 & 3)   # operação bit a bit em inteiros
print('5 | 3 =', 5 | 3)   # operação bit a bit em inteiros
````

## 1. Aprovação em uma disciplina
Um aluno é aprovado se a nota for maior ou igual a 7 **e** a frequência for maior
ou igual a 75%.

````python
nota = float(input('Digite a nota: '))
freq = int(input('Digite a frequência: '))

if nota >= 7 and freq >= 75:
    print('Aprovado')
else:
    print('Reprovado')

print('Nota:', nota)
print('Frequência: ', freq)


````

## 2. Desconto em uma loja

Uma loja dá desconto especial para clientes que forem **maiores de idade**
(18 anos ou mais) **e** que sejam **clientes cadastrados**.

Peça para o usuário digitar a **idade** e informar se ele é **cliente cadastrado**
(por exemplo, digitando "s" para sim ou "n" para não). Verifique se ele tem
direito ao desconto e exiba a mensagem correspondente.

**Dica:** para transformar a resposta "s"/"n" em um valor booleano, você pode
comparar a entrada com a string esperada, por exemplo:
```python
cadastrado = input('É cliente cadastrado? (s/n): ') == 's'
```

````python
idade = int(input('Digite sua idade: '))
cadastrado = input('É cliente cadastrado? (s/n): ') == 's'

if idade >= 18 and cadastrado:
    print('Parabéns, você tem direito ao desconto!')
elif idade >= 18 and not cadastrado:
    print('Faça nosso cadastro e garanta seu desconto!')
else:
    print('Desculpe, você não tem direito ao desconto.')
````

## 3. Acesso a um sistema

O acesso a um sistema deve ser liberado se o usuário digitar o **login correto**
**e** a **senha correta**, **ou** se ele informar um **token de administrador válido**
(que dá acesso mesmo sem login/senha corretos).

Considere como corretos:
- Login: `admin`
- Senha: `1234`
- Token de administrador: `master2026`

Peça para o usuário digitar **login**, **senha** e **token** (o usuário pode deixar
o token em branco se não tiver um). Verifique se o acesso deve ser liberado e exiba
a mensagem correspondente.

**Dica:** você vai precisar combinar `and` com `or`. Fique atento à ordem de
precedência entre eles — se tiver dúvida, use parênteses para deixar explícito
o que deve ser avaliado primeiro:
```python
acesso = (login == login_correto and senha == senha_correta) or token == token_correto
```

````python
loginCor = 'admin'
senhaCor = '1234'
tokenCor = 'master2026'

login = input('Digite seu login: ')
senha = (input('Digite sua senha: '))
token = input('Caso seja um administrador digite seu token: ')

acesso = (login == loginCor and senha == senhaCor) or token == tokenCor

if acesso:
    print('Acesso liberado')
else:
    print('Acesso negado')


````

## 4. Empréstimo bancário

Um banco aprova um empréstimo se a **renda for maior que R$ 3000** **e** o
cliente **não estiver com o nome negativado**.

Peça para o usuário digitar a **renda mensal** e informar se está **negativado**
(por exemplo, "s" para sim ou "n" para não). Verifique se o empréstimo deve ser
aprovado e exiba a mensagem correspondente.

**Dica:** lembre-se do operador `not` para inverter a condição de estar negativado.

````python
renda = float(input('Digite sua renda mensal: '))
negativado = input('Está negativado? (s/n): ') == 's'

aprovado = renda > 3000 and not negativado

if aprovado:
    print('O seu crédito foi aprovado')
else:
    print('Crédito negado')
````
