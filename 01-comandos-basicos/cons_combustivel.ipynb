#Script em Python que recebe a distância percorrida por um carro, a quantidade de litros consumidos nessa distância, uma nova distância a ser percorrida e exibe quantos litros serão necessários para percorrer essa nova distância.

##A distância percorrida e o consumo do veículo são informados pelo usuário:

```python
distanciaPercorrida = float(input('Qual a distancia percorrida em km? '))
```

```python
consumoVeiculo = float(input('Quantos litros de combustível foram consumidos? '))
```

```python
kmPorLitro = (distanciaPercorrida) / (consumoVeiculo)
```

```python
print(kmPorLitro)
```

##O usuário informa qual distância ainda precisa percorrer:

```python
novaDistancia = float(input('Qual distancia ainda será percorrida? '))
```

```python
consumoFinal = (novaDistancia) / (kmPorLitro)
```

##Litros necessários para percorrer a nova distância.

```python
print(f'Para percorrer {novaDistancia} km, serão necessários {consumoFinal} litros de combustível.')
```
