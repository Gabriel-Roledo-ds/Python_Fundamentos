```python
#exercicio 3
'''Receba dois números inteiros, faça e exiba: a potência entre os dois
números, o resto da divisão entre os dois números e a divisão
inteira entre os dois números.'''

#entrada
numero1 = int(input('Digite um numero: '))
numero2 = int(input('Digite outro numero: '))

#processamento
potenc = numero1 ** numero2
restoDivi = numero1 % numero2
divIntei = numero1 // numero2

#saída

print(f'O primeiro numero elevado ao segundo numero é: {potenc}')
print(f'O resto da divisão do primeiro numero pelo segundo numero é: {restoDivi}')
print(f'A divisão inteira do primeiro numero pelo segundo numero é: {divIntei}')
```

```python
#exercicio 4
'''Receba um número inteiro e calcule sua raiz quadrada e sua raiz
cúbica.'''


#entrada
numero1 = int(input('Digite um numero: '))

#processamento

raizQuad = numero1 ** (1/2)
raizCub = numero1 ** (1/3)

#saída

print(f'A raiz quadrada do numero digitado é: {raizQuad}')
print(f'A raiz cubica do numero digitado é: {raizCub:.2f}')

```

```python
#exercicio 5
'''Teste o seguinte trecho de código abaixo e verifique as formas
diferentes de saídas:'''


#entrada
nome = input('Digite seu nome: ')
print(f"Bem vindo {nome}!")
print(f"Bem vindo {nome:20}!")
print(f"Bem vindo {nome :>20}!")
print(f"Bem vindo {nome:<20}!")
print(f"Bem vindo {nome:^20}!")
print(f"Bem vindo {nome:=^20}!")
```
