#Script em Python que recebe o valor total de vendas realizadas por um vendedor, o percentual de comissão e exibe o valor da comissão a ser recebida.

###As vendas realizadas e o percentual de comissão são informadas pelo usuário:

```python
vendasRealizadas = float(input("Digite o valor total de vendas realizadas: R$"))
percentualComissao = float(input("Digite o percentual de comissão: "))
```

###Cálculo da comissão a ser recebida:

```python
comissao = (vendasRealizadas) * (percentualComissao / 100)
```

```python
print(f'A comissão a ser recebida pelo vendedor é de R${comissao:.2f}.')
```
