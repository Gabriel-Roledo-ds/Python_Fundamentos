```python
#exercicio 5
'''Considere a função f(x) = 2x + 3
Entrada: O usuário deve digitar o valor de x.
Processamento: Substituir o valor de x na função.
Saída: Exibir o valor de f(x).'''

x = float(input('Digite o valor de x: '))

#processamento

y = 2 * x + 3

#saída

print(y)
```

```python
#exercicio 6
'''Considere a função f(x) = ax + b
Entrada: O usuário deve digitar os valores de a, b e x.
Processamento: Calcular o valor da função.
Saída: Exibir o valor calculado de f(x).'''


x = float(input('Digite o valor de x: '))
a = float(input('Digite o valor de a: '))
b = float(input('Digite o valor de b: '))

#processamento

y = a * x + b

#saída

print(y)
```

```python
#exercicio 7
'''Entrada: O usuário deve digitar o salário base e o percentual de bônus.
Processamento: Calcular o bônus e somar ao salário base.
Saída: Exibir o valor do bônus e o salário final.'''

#entrada
salarioBase = float(input('Digite o valor do salário base: '))
percBonus = float(input('Digite o valor do percentual de bonus: '))

#processamento

bonus = salarioBase * percBonus /100
salarioFinal = salarioBase + bonus

#saída

print(f'O valor do bônus é de R${bonus:.2f} e o salário final é de R${salarioFinal:.2f}')
```

```python
#exercicio 8
'''Considere a função d = vt + s
Entrada: O usuário deve digitar velocidade, tempo e posição inicial.
Processamento: Calcular a distância final.
Saída: Exibir a distância calculada.'''

#entrada
veloc = float(input('Digite a velocidade que correu (km/h): '))
tempo = float(input('Digite o tempo que levou (h): '))
posicaoInic = float(input('Digite a posição inicial: '))

#processamento
distFinal = veloc * tempo + posicaoInic

#saída
print(distFinal)
```

```python
#exercicio 9
'''Considere a função valor = 5x + 10
Entrada: O usuário deve digitar a quantidade de quilômetros percorridos.
Processamento: Calcular o valor da corrida.
Saída: Exibir o valor total a pagar.'''

#entrada
kmPerc = float(input('Digite a quantidade de quilômetros percorridos: '))

#processamento
valor = 5 * kmPerc + 10

#saída
print(f'O valor total a pagar é de R${valor:.2f}')
```

```python
# Exercício 10

'''Considere a função C(x) = 0.12x + 3
Entrada: O usuário deve digitar a quantidade de quilômetros percorridos.
Processamento: Calcular o consumo estimado.
Saída: Exibir o consumo total.'''

kmPerc = float(input('Digite a quantidade de quilômetros percorridos: '))

#processamento
consumo = 0.12 * kmPerc + 3

#saída
print(f'O consumo estimado é de {consumo:.2f} litros')
```
