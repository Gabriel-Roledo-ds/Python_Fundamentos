#Script em Python que recebe a nota da primeira prova (peso de 40%) e a nota da segunda prova (peso de 60%) e exibe a média final ponderada do aluno.

###As notas 1 e 2 são informadas pelo usuário:

```python
nota1 = float(input('Digite a nota da primeira prova: '))
nota2 = float(input('Digite a nota da segunda prova: '))
```

###Cálculo da média:

```python
mediaFinal = (nota1 * 0.4) + (nota2 * 0.6)

print(f'A média final ponderada do aluno é: {mediaFinal:.2f}')
```
