#Script em Python que recebe o valor de um produto , o percentual de desconto e o valor final do produto.

Valor do produto:

```python
valorProduto = float(input('Digite o valor do produto: '))
```

Valor do desconto:

```python
valorDesconto = float(input('Digite o percentual do desconto: '))
```

```python
desconto = (valorProduto) * (valorDesconto / 100)
```

```python
print (f'O valor do desconto é de {desconto}%')
```

```python
valorFinal = (valorProduto) - (desconto)
```

Resultado:

```python
print(f'O valor do desconto é de {valorDesconto}% e o valor final do produto é de R${valorFinal}.')
```
