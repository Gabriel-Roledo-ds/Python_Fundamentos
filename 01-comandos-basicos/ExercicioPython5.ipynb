#Script em Python que recebe o consumo de energia elétrica de uma residência (em Kwh) no mês anterior, o percentual de redução no consumo, e exibe a quantidade de energia economizada e o novo consumo.

##O consumo de energia   e o percentual de consumo são informados pelo usuário:

```python
consumoAnterior = float(input('Digite o consumo de energia do mês anterior em kWh: '))
percentualReducao = float(input('Digite o percentual de redução no consumo: '))
```

##Cálculo da energia economizada:

```python
energiaEconomizada = (consumoAnterior) * (percentualReducao / 100)
novoConsumo = consumoAnterior - energiaEconomizada

print(f'A energia economizada foi de {energiaEconomizada:.2f} kWh e o novo consumo foi de {novoConsumo:.2f} kWh')
```
