#Script em Python que recebe o salário, o percentual de aumento e o valor do novo salário de um funcionário.

### O valor inicial do salário e o percentual de aumento é informado pelo usuário:



```python
salario = float(input('Digita o valor do salário: '))
```

```python
valorAumento = float(input('Digite o percentual do aumento: '))
```

###Cálculo do aumento salarial:

```python
aumento = (salario) * (valorAumento/100)
```

###Valor final do salário:

```python
salarioFinal = (salario) + (aumento)
```

```python
print (f'O aumento salarial é de {valorAumento}% e o novo salário é de {salarioFinal}')
```
